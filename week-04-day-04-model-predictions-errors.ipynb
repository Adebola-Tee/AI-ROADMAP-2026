```python
import pandas as pd
```

```python
data = {
    "student": ["Achi", "Ochi", "Muchi"],
    "homework_score": [55, 68, 83],
    "mock_exam_score": [72, 78, 94],
    "participation_score": [82, 56, 92],
    "actual_final_score": [70, 74, 93],
    "predicted_final_score": [68, 76, 90]
}
df = pd.DataFrame(data)
df
```

```python
df["error"] = df["actual_final_score"] -  df["predicted_final_score"]
df
```

```python
df["absolute_error"] = df["error"].abs()
df
```

```python
average_absolute_error = df["absolute_error"].mean()
print(f"The average absolute error is: {average_absolute_error:.2f}")
```

```python
largest_error = df[df["absolute_error"]== df["absolute_error"].max()]
largest_error
```
