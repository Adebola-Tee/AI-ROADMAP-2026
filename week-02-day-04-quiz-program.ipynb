# 3-question quiz program.

```python
import time
score = 0
total_score = 3


q1 = int(input ("What year are we currently in?"))
if q1 != 2026:
    print ("Incorrect answer!")
else:
    print ("Correct answer!")
    score +=1
    print ("You have scored yourself 1 point")
q2 = input("Is Nigeria in Africa?")
if q2 != "yes":
    print ("Incorrect answer!")
else:
    print ("Correct answer!")
    score +=1
    print ("You have scored yourself 1 point")
q3 = int(input ("How many days are there in a leap year?"))
if q3 != 366:
    print ("Incorrect answer!")
else:
    print ("Correct answer!")
    score +=1
    print ("You have scored yourself 1 point")
print ("-----------------GAME OVER---------------------")
time.sleep(10)
if score == 1:
    print(f"You have answered 1 correct answer out of {total_score}")
elif score == 2:
    print(f"You have answered 2 correct answers out of {total_score}")
elif score == total_score:
    print("Horray!You answered all questions")

    
    
```
