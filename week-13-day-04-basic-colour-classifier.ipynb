# Basic Color Identifier

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
red_image = np.zeros((5, 5, 3), dtype=np.uint8)

red_image[:, :, 0] = 255
```

```python
green_image = np.zeros((5, 5, 3), dtype=np.uint8)

green_image[:, :, 1] = 255
```

```python
blue_image=np.zeros((5,5,3), dtype=np.uint8)
blue_image[:,:,2]=255
```

```python
red_average=red_image[:,:,0].mean()
green_average=red_image[:,:,1].mean()
blue_average=red_image[:,:,2].mean()
```

```python
print("Red:", red_average)
print("Green:", green_average)
print("Blue:", blue_average)
```

```python
def classify_colour(image):
    red_average=image[:,:,0].mean()
    green_average=image[:,:,1].mean()
    blue_average=image[:,:,2].mean()
    if red_average>green_average and red_average>blue_average:
        return"Red"
    elif green_average>red_average and green_average>blue_average:
        return "Green"
    else:
        return "Blue"
    
```

```python
print (classify_colour(blue_image))
```

```python
mostly_red = np.zeros((5, 5, 3), dtype=np.uint8)

mostly_red[:, :, 0] = 220
mostly_red[:, :, 1] = 40
mostly_red[:, :, 2] = 30
```

```python
plt.imshow(mostly_red)
plt.show
```

```python
print(classify_colour(mostly_red))
```

```python
my_image=np.zeros((5,5,3), dtype=np.uint8)
my_image[:,:,0]=20
my_image[:,:,1]=10
my_image[:,:,2]=230
```

```python
print(classify_colour(my_image))
```

```python
def analyse_colour(image):
    red_average=image[:,:,0].mean()
    green_average=image[:,:,1].mean()
    blue_average=image[:,:,2].mean()
    
    prediction=classify_colour(image)

    print("Average Red:", red_average)
    print("Average Green:", green_average)
    print("Average Blue:", blue_average)
    print("Prediction:", prediction)
```

```python
analyse_colour(mostly_red)
```
