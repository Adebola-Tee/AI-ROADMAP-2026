## Today, I am learning how computer represents and manipulate images as numerical tensors

```python
import numpy as np
import matplotlib.pyplot as plt

image = np.array([
    [0,   0,   0,   0,   0],
    [0, 255, 255, 255,   0],
    [0, 255,   0, 255,   0],
    [0, 255, 255, 255,   0],
    [0,   0,   0,   0,   0]
])

image
```

## This is not yet a picture to Python in the human sense; it is a 2D array.

```python
print("Shape:", image.shape)
print("Dimension", image.ndim)
print("Type:", image.dtype)
```

## Shape: Size of each dimension, ndim: Number of dimensions, dtype: Type of numbers stored inside.

```python
plt.imshow(image, cmap="gray")
plt.show
```
