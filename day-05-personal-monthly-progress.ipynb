# Personal Monthly Progress Calculator (Input Version)

```python
monthly_income = float(input("Enter monthly income:"))
monthly_spending = float(input("Enter monthly spending:"))
money_left_after_spending = monthly_income - monthly_spending
percentage_of_income_saved = (money_left_after_spending / monthly_income ) *100
money_to_save_in_three_months = money_left_after_spending * 3
money_to_save_in_six_months = money_left_after_spending * 6
current_saving_goal = float(input("Enter current saving plan:"))
money_remaining_to_achieve_goals = current_saving_goal -  money_to_save_in_six_months
print(f"Money left after spending is {money_left_after_spending}.")
print(f"Percentage of income saved is {percentage_of_income_saved}%.")
print(f"How much you could save in 3 months is £{money_to_save_in_three_months}.")
print(f"How much you could save in 6 months is £{money_to_save_in_six_months}.")
print(f"Money remaining to achieve the goal £{money_remaining_to_achieve_goals}")
```
