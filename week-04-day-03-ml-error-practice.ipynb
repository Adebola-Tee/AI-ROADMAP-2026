# Week 4 Day 3 — ML Error Practice

Today, I am practising how to understand features, target values, predictions, errors, and a simple loss function.

```python
import pandas as pd
data = {
    "student": ["A", "B", "C"],
    "study_hours": [2, 4, 6],
    "attendance": [70, 80, 90],
    "actual_score": [48, 65, 82],
    "predicted_score": [50, 62, 78]
}

df = pd.DataFrame(data)
df
```

```python
df["error"] = df["actual_score"] - df["predicted_score"]
df
```

### Question 1: Which columns are the features?

The feature columns are `study_hours` and `attendance`.

These are the pieces of information given to the model to help it make a prediction.

### Question 2: Which column is the target?

The target column is `actual_score`.

This is the real answer the model is trying to predict.

### Question 3: What does each row represent?

Each row represents one student.

For each student, the table shows their study hours, attendance, actual score, and predicted score.

### Question 4 — Calculate each student’s prediction error
df["error"] = df["actual_score"] - df["predicted_score"]
df

### Question 5 — Which prediction was closest?

```python
df["absolute_error"]= df["error"].abs()
df
```

```python
closest_prediction = df[df["absolute_error"] == df["absolute_error"].min()]
closest_prediction
```

### Question 6 — Which prediction was furthest away?

```python
furthest_prediction = df[df["absolute_error"] == df["absolute_error"].max()]
furthest_prediction
```

### Question 7 — Calculate the average actual score

```python
average_actual_score = df["actual_score"].mean()
print(f"The average actual score is {average_actual_score:.2f}")
```

### Question 8 — Calculate the average predicted score

```python
average_predicted_score = df["predicted_score"].mean()
print(f"The average predicted score is {average_predicted_score:.0f}")
```
