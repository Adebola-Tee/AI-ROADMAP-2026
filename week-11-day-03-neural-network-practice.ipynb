# Exercise 1 — Identify the inputs
## inputs are:
## study_hours
## attendance
## homework_score

```python
study_hours = 5
attendance = 0.85
homework_score = 0.75
```

# Exercise 2 — Identify the weights


```python
study_weight = 0.1
attendance_weight = 0.2
homework_weight = 0.8
```

## Which input has the greatest weight?
## Study_hours
## The larger weight means that the network is giving study_hours more importance.

# Exercise 3 — Calculate each weighted input separately

```python
weighted_study = study_hours * study_weight
weighted_attendance = attendance * attendance_weight
weighted_homework = homework_score * homework_weight

print(weighted_study)
print(weighted_attendance)
print(weighted_homework)
```

## Weighted input = input * weight

# Exercise 4 — Combine the weighted inputs

```python
combined_value = (
    weighted_study
    + weighted_attendance
    + weighted_homework
)

print(combined_value)
```

## So, this is what the neuron has gathered before the bias and activation.

# Exercise 5 — Meet bias properly

```python
bias = 0.5
```

```python
combined_with_bias = combined_value + bias
print(combined_with_bias)
```

## In summary, bias is an extra adjustment added to a neuron's combined value.

# Exercise 6 — A very simple activation function

```python
def simple_activation(value):
    if value > 2:
        return 1
    else:
        return 0
        
```

```python
output = simple_activation(combined_with_bias)

print(output)
```

## 1 = neuron responds strongly
## 0 = neuron does not respond strongly
## Activation takes the neuron's combined value and transforms it to a response.
