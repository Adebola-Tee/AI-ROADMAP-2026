# Student Grade Predictor


```python
Student_name =  input ("Enter your name:")
score = int(input("Enter your score:"))
if 0<= score <=49:
    print (f"{Student_name}, your score {score} needs improvement")
elif 50<= score <=59:
     print (f"{Student_name}, your score {score} was good")
elif 60<= score <=69:
     print (f"{Student_name}, your score {score} pass")
elif 70<= score <=79:
     print (f"{Student_name}, your score {score} very good")
elif 80<= score <=100:
     print (f"{Student_name}, your score {score} excellent")
```
