# Images as Tensors, Classification, and the idea of transfer learning.

## A computer sees an image as numbers.
## An image is made from a lot of tiny squares called PIXELS
0 means very dark.
255 means very bright.
## A tensor is basically a structure that can hold numbers in multiple directions

## Single number -- scalar
## list of numbers - 1D
## Rows and Columns - 2D
## Rows * Columns - multiple dimensions

## For a color image, we have height, width, and channels
## For example 100 * 100 * 3
## That means 100 pixels high, 100 pixels wide, and 3 color channels
## The 3 represent Red, Green, and Blue—so a color picture is essentially made by combining these 3 channels
## An image tensor is a numerical representation of an image

## Image classification - looks at an image and decide what category it belongs to.

## Differences between classification and object detection
## Image classification—"What is in this image?"
## Object Detection—"What objects are here AND where are they?"
## For example: Cat at this position, dog at this position

## A computer vision uses a special kind of neural network called CNN (Convolutional Neural Network)

## Transfer Learning—Instead of training a neural network from scratch, we can take a model that has learnt useful visual patterns from a very large image dataset. Then we adapt it to our own problem.

## Transfer Learning—Reuse the knowledge learned to solve one problem to help another related problem.


# Pretrained Model—A pretrained model is simply a model that someone has already trained.
# Transfer learning can save time, computing power, and training data.
