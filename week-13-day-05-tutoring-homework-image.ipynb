```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

## PIL - Python Imaging Library

```python
student_image=Image.open("student_work.jpeg")
```

```python
student_image=student_image.convert("RGB")
```

## Convert the image into numbers

```python
image_array=np.array(student_image)
```

```python
image_array
```

```python
print(image_array.shape)
```

## Check its dimension

```python
print("Dimensions:", image_array.ndim)
print("Data type:", image_array.dtype)
```

```python
plt.imshow(image_array)
plt.show()
```

```python
image_array[0,0]
```

```python
normalised_image=image_array/255.0
```

```python
print(normalised_image.min())
print(normalised_image.max())
```

```python
plt.imshow(normalised_image)
plt.show()
```

## Recall that normalisation changes number used by the model, not the meaning of the image

```python
red_channel = image_array[:, :, 0]
green_channel = image_array[:, :, 1]
blue_channel = image_array[:, :, 2]
```

```python
print(red_channel.shape)
print(green_channel.shape)
print(blue_channel.shape)
```

## Calculate average channel strengths

```python
red_average=red_channel.mean()
green_average=green_channel.mean()
blue_average=blue_channel.mean()
print("Average Red:", red_average)
print("Average Green:", green_average)
print("Average Blue:", blue_average)
```
