# Student Readiness Neuron

```python
study_hours = 7
attendance = 10
homework_score = 90
```

```python
study_weight = 0.4
attendance_weight = 0.35
homework_weight = 0.25
```

```python
bias = 0.3
```

```python
weighted_sum = (
    study_hours * study_weight
    + attendance * attendance_weight
    + homework_score * homework_weight
)

neuron_value = weighted_sum + bias
```

```python
def readiness_activation(value):
    if value > 2.5:
        return 1
    else:
        return 0
```

```python
output = readiness_activation(neuron_value)
print(output)
```
