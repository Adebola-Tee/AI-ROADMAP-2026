```python
import pandas as pd
df= pd.read_csv("student_final_score_project.csv")
df
```

```python
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].mean())
df
```

```python
df.isnull().sum()
```

```python
df["attendance"] = df["attendance"].fillna(df["attendance"].mean())
df
```

```python
df["homework_score"] = df["homework_score"].fillna(df["homework_score"].mean())
df
```

```python
df["previous_test_score"] = df["previous_test_score"].fillna(df["previous_test_score"].mean())
df
```

```python
df["tutoring_sessions"] =df["tutoring_sessions"].fillna(df["tutoring_sessions"].mean())
df
```

```python
df["sleep_hours"] = df["sleep_hours"].fillna(df["sleep_hours"].mean())
df
```

```python
df["final_score"] = df["final_score"].fillna(df["final_score"].mean())
df
```

```python
df.isnull().sum()
```

```python
X = df[[
        "study_hours",
        "attendance",
        "homework_score",
        "previous_test_score",
        "tutoring_sessions",
        "sleep_hours"
    ]]
y = df["final_score"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.3,
    random_state =42
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
predictions
```

```python
compare_final_score = pd.DataFrame ({
    "Actual_Score": y_test.values,
    "Predicted_Score": predictions
})
compare_final_score
```

# It's time to evaluate with MAE and R-SQUARED

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
mae = mean_absolute_error(y_test, predictions)
print(mae)
```

```python
r2 = r2_score(y_test, predictions)
print(r2)
```
