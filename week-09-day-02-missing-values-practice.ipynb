```python
import pandas as pd
import numpy as np
```

```python
data = {
    "student":["A", "B", "C", "D", "E"],
    "study_hours":[2, 4, np.nan, 6, 5],
    "attendance":[70, np.nan, 85, 90, 88],
    "score":[50, 65, 72, np.nan, 80]
}
df= pd.DataFrame(data)
df
```

## np.nan means missing value

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].mean())
df["study_hours"]
```

```python
df["attendance"]= df["attendance"].fillna(df["attendance"].mean())
df["attendance"]
```

```python
df["score"] = df["score"].fillna(df["score"].mean())
df["score"]
```

### For the missing values in study_hours, attendance, and score, I found the mean and used it to fill the blank in each.
