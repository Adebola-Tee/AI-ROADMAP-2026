```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    "study_hours": [1.5, 2.0, 1.0, 2.5, 3.0],
    "tasks_completed": [2, 3, 1, 4, 5],
    "focus_score": [6, 7, 5, 8, 9]
}

df = pd.DataFrame(data)
df
```

```python
df["study_hours"].sum()
df["study_hours"].mean()
df["tasks_completed"].sum()
df["focus_score"].mean()
df.loc[df["study_hours"].idxmax()]
df.loc[df["focus_score"].idxmax()]
```

```python
plt.bar(df["day"], df["study_hours"])
plt.title("AI Study Hours by Day")
plt.xlabel("Day")
plt.ylabel("Study Hours")
plt.show()
```

```python
plt.plot(df["day"], df["tasks_completed"], marker="o")
plt.title("Tasks Completed by Day")
plt.xlabel("Day")
plt.ylabel("Tasks Completed")
plt.show()
```

```python
plt.scatter(df["study_hours"], df["focus_score"])
plt.title("Study Hours and Focus Score")
plt.xlabel("Study Hours")
plt.ylabel("Focus Score")
plt.show()
```
