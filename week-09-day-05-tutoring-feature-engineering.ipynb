```python
import pandas as pd
import numpy as np

data = {
    "student": ["A", "B", "C", "D", "E", "F"],
    "study_hours": [2, 4, np.nan, 5, 6, 7],
    "attendance": [70, 78, 82, np.nan, 91, 95],
    "subject": ["Maths", "English", "Maths", "English", np.nan, "Maths"],
    "homework_score": [50, 62, 70, 75, np.nan, 90],
    "final_score": [48, 60, 69, np.nan, 85, 92]
}

df = pd.DataFrame(data)
df
```

```python
copy_df=df.copy()
copy_df
```

```python
copy_df["study_hours"] = copy_df["study_hours"].fillna(copy_df["study_hours"].mean())
copy_df
```

```python
copy_df["attendance"]=copy_df["attendance"].fillna(copy_df["attendance"].mean())
copy_df
```

```python
copy_df["subject"] = copy_df["subject"].fillna(copy_df["subject"].mode()[0])
copy_df
```

```python
copy_df["homework_score"] = copy_df["homework_score"].fillna(copy_df["homework_score"].median())
copy_df["homework_score"]
```

```python
copy_df["final_score"] = copy_df["final_score"].fillna(copy_df["final_score"].median())
copy_df
```
