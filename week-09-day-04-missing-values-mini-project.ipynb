## CLEANING A MESSY STUDENT DATASET TO PREPARE IT FOR MACHINE LEARNING

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "student": ["A", "B", "C", "D", "E", "F"],
    "study_hours": [2, 4, np.nan, 5, 6, 8],
    "attendance": [70, np.nan, 82, 90, 95, 88],
    "subject": ["Maths", "English", "Maths", np.nan, "Maths", "English"],
    "score": [50, 65, 72, np.nan, 88, 91]
}
df=pd.DataFrame(data)
df
```

```python
clean_df = df.copy()
clean_df
```

```python
clean_df["study_hours"]= clean_df["study_hours"].fillna(clean_df["study_hours"].mean())
clean_df
```

```python
clean_df["attendance"]=clean_df["attendance"].fillna(clean_df["attendance"].mean())
clean_df
```

```python
clean_df["subject"]=clean_df["subject"].fillna(clean_df["subject"].mode()[0])
clean_df
```

```python
clean_df["score"]=clean_df["score"].fillna(clean_df["score"].median())
clean_df
```
