# Weekly Tutoring Data Analysis

## 1. Project Introduction
This project analyses a fictional weekly tutoring dataset to understand lessons, hours worked, income patterns, and simple insights.

## 2. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    "lessons": [3, 5, 2, 4, 6],
    "hours_worked": [3.0, 4.5, 2.0, 4.0, 5.5],
    "income": [18000, 30000, 12000, 24000, 36000]
}

df = pd.DataFrame(data)
df
```

```python
df.head()
df.info()
df.describe()
```

```python
total_income = df["income"].sum()
average_income = df["income"].mean()
total_lessons = df["lessons"].sum()
average_hours = df["hours_worked"].mean()

print(f"Total income: ₦{total_income}")
print(f"Average daily income: ₦{average_income:.2f}")
print(f"Total lessons: {total_lessons}")
print(f"Average hours worked: {average_hours:.2f}")
```

```python
plt.bar(df["day"], df["income"])
plt.title("Tutoring Income by Day")
plt.xlabel("Day")
plt.ylabel("Income (₦)")
plt.show()
```

```python
plt.plot(df["day"], df["income"], marker="o")
plt.title("Daily Tutoring Income")
plt.xlabel("Day")
plt.ylabel("Income (₦)")
plt.show()
```

```python
plt.scatter(df["hours_worked"], df["income"])
plt.title("Hours Worked and Income")
plt.xlabel("Hours Worked")
plt.ylabel("Income (₦)")
plt.show()
```

## Insights

1. Friday produced the highest income at ₦36,000.
2. Wednesday had the fewest lessons and the lowest income.
3. Income generally increased as hours worked increased.
