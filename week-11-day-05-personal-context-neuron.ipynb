```python
attendance = 0.82
homework_score = 0.74
previous_test_score = 0.68
```

```python
attendance_weight = 0.4
homework_weight = 0.35
previous_test_weight = 0.25
```

```python
weighted_value = (
    attendance * attendance_weight
    + homework_score * homework_weight
    + previous_test_score * previous_test_weight
)
```

```python
bias = 0.2
```

```python
combined_value = weighted_value + bias
```

```python
def simple_activation(value):
    if value > 0.75:
        return 1
    else:
        return 0
```

```python
output = simple_activation(combined_value)
print(output)
```
