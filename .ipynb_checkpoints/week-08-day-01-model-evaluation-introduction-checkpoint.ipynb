# Classical ML Models Recaps
# Linear Regression—For predicting a number by learning a numerical relationship
# For example: Study_hours -> Final_score
# Logistic Regression - Predict a category
# For example: Pass/Fail, Yes/No, Spam/No Spam
# Decision Tree: Makes predictions by asking a sequence of questions
# Random Forest: Using many decision trees instead of trusting only one decision tree. Each tree gives an answer, and then a random forest combines those answers.
# K-Nearest Neighbours(KNN): Looking for past examples that are most similar to the new one.
# Example: A new student studies 5 hours and has 85% attendance. KNN looks for previous students with similar study hours and attendance, then uses their outcomes to help make the prediction


# Model Evaluation: After building a model, the next question is, "How good is the model?"
# Accuracy: Accuracy asks, 'Out of everything the model predicted, how many predictions are correct?"
# For example: 10 predictions, 8 of them are correct, so 80% accuracy. So, accuracy is overall correctness.
# Precision: Precision asks, "When the model says YES, how often is it actually right?"
# For example: The model predicts that 5 students will pass. Only 4 actually pass. Precisions looks at those 5 predictions and asks how many are correct. So, precision is about 'Can I trust the model when it says YES?'
# Recall: Recall asks, 'Out of all the real YES cases, how many did the model successfully find?'
# For example: 6 students actually passed. The model correctly identified only 4 of them. Recall asks how many of the real passes were found.
# F1 score: It balances both precision and recall.
# Confusion Matrix: It shows the model's correct predictions and mistakes. It is more like the model's detailed report card.
# Overfitting: This is when a model performs too good on training data but performs poorly on new data. it is like a student who memorizes the exact past questions instead of understanding the subject. In summary, overfitting is memorizing instead of learning.
