```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    "lessons": [3, 5, 2, 4, 6],
    "hours_worked": [3.0, 4.5, 2.0, 4.0, 5.5],
    "income": [18000, 30000, 12000, 24000, 36000]
}

df = pd.DataFrame(data)
df
```

### Total Income

```python
 df["income"].sum()
```

### Average daily income

```python
df["income"].mean()
```

### Total lessons

```python
df["lessons"].sum()
```

### Day with the highest income

```python
df.loc[df["income"].idxmax()]
```

### Days with at least 4 lessons

```python
df[df["lessons"] >= 4]
```

### Create Charts

### Line chart

```python
plt.plot(df["day"], df["income"], marker="o")
plt.title("Daily Tutoring Income")
plt.xlabel("Day")
plt.ylabel("Income ($)")
plt.show()
```

### Scatter Plot

```python
plt.scatter(df["hours_worked"], df["income"])
plt.title("Hours Worked and Income")
plt.xlabel("Hours Worked")
plt.ylabel("Income ($)")
plt.show()
```
