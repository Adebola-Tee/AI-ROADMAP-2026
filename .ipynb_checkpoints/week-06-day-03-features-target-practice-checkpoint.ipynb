# Exercise 1 — Student Score Prediction
# This is a regression - predicting a number

```python
import pandas as pd
data = {
    "study_hours": [2,4,6,8],
    "attendance":[60,75,90,95],
    "sleep_hours":[8,7,6,5],
    "score":[45,64,85,92]
}
df=pd.DataFrame(data)
df
```

```python
X =df[["study_hours", "attendance", "sleep_hours"]]
y= df["score"]
print("Features:")
print(X)
print("Target:")
print(y)
```

# Exercise 2 — House Price Prediction

```python
info={
"bedrooms":[1,2,3,4],
"bathrooms":[1,2,3,4],
"house_size":[100,150,200,250],
"location_score":[32,3,4,5],
"price":[5000,7500,9000,11500]
}
df=pd.DataFrame(info)
df
```

```python
X =df[["bedrooms", "bathrooms", "house_size", "location_score"]]
y=df["price"]
print("Features:")
print(X)
print("Target:")
print(y)
```

# Exercise 3 — Pass or Fail Prediction
# This is classification = prediction of category

```python
class_info={
"study_hours": [2,4,6,8],
"attendance":[60,75,90,95],
"assignment_score": [45,64,85,92],
"result":["fail", "pass", "pass", "pass"]
}
df=pd.DataFrame(class_info)
df
```

```python
X = df[["study_hours", "attendance", "assignment_score"]]
y=df["result"]
print("Features:")
print(X)
print("Targer:")
print(y)
```
