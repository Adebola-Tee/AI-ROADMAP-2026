# Daily Task Tracker

```python
Task_Tuesday = ["Brush my teeth", "Practice IELTS"]
Task_Tuesday.append("Pray")
Task_Tuesday.append("Eat")
Task_Tuesday.append("Update website")
Task_completed = input ("what task did you complete?")
if Task_completed in Task_Tuesday:
    print (f"Well done! you have completed {Task_completed}")
else:
    print(f"Oops! {Task_completed} is not present in {Task_Tuesday}")
```
