# Day 39 — Student Performance Dataset Preparation

This mini-project practises how to prepare a simple dataset for machine learning.

The goal is to identify the features, identify the target, and separate the dataset into `X` and `y`.

```python
import pandas as pd

data = {
    "study_hours": [2, 4, 6, 8, 3],
    "attendance": [60, 75, 90, 95, 70],
    "homework_score": [50, 68, 85, 92, 60],
    "score": [45, 64, 85, 92, 58]
}

df = pd.DataFrame(data)
df
```

## Identify Features and Target

The features are the columns that will help the model make a prediction.

The target is the answer the model is trying to predict.

X = df[["study_hours", "attendance", "homework_score"]]

y = df["score"]

print("Features:")
print(X)

print("Target:")
print(y)

## Task Type

This is a regression problem.

It is regression because the model is trying to predict a number: the student's `score`.

## Conclusion

In this mini-project, I created a small student performance dataset.

I identified the feature columns as `study_hours`, `attendance`, and `homework_score`.

I identified the target column as `score`.

I also separated the data into `X` and `y`, which is an important step before training a machine-learning model.
