# Student Pass/Fail Model Evaluation

```python
actual =    [1, 0, 1, 1, 0, 1, 0, 1, 0, 1]
predicted = [1, 0, 1, 0, 0, 1, 1, 1, 0, 0]
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
accuracy = accuracy_score(actual, predicted)
precision = precision_score(actual, predicted)
recall = recall_score(actual, predicted)
f1 = f1_score(actual, predicted)

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
cm = confusion_matrix(actual, predicted)
cm
```

## My Interpretation
1. How many predictions were correct overall? - 4

2. How many students were False Positives?
   Meaning: model predicted Pass, but student actually Failed. - 2

3. How many students were False Negatives?
   Meaning: model predicted Fail, but student actually Passed. - 1

4. Based on the metrics, do I think this model performed reasonably well? YES
