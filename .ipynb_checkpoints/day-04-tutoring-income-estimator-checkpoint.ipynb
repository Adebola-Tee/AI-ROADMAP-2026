# DAY 4 -Tutoring Monthly Income Estimator


```python
number_of_students = 5
fee_per_lesson = 200
number_of_classes_per_student_per_week = 1
number_of_classes_per_month = number_of_students * number_of_classes_per_student_per_week * 4
earning_per_month = fee_per_lesson * number_of_classes_per_month 
earning_after_platform_charge = earning_per_month - (earning_per_month * 0.25)
print (f"My earnings before the platform fee charges is {earning_per_month } HKD.")
print (f"Platform fee charges is {earning_per_month * 0.25} HKD.")
print (f"My earnings after the platform fee charges is {earning_after_platform_charge} HKD.")
```
