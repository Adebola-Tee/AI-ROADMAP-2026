# Feature: This is the information given to the model
# Label/Target: This is the answer the model is trying to predict
# Training data: This is the data the model is trying to learn from
# Testing data: This is the new data used to check whether the model learnt well
# Prediction: This is the model's estimated answer
# Regression: This means predicting a number
# Classification: This means predicting a category
# Model evaluation: This means checking how well the model performed
