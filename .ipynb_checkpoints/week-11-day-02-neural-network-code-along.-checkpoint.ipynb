```python
study_hours = 6
attendance = 0.9
homework_score = 0.8

study_weight = 0.5
attendance_weight = 0.3
homework_weight = 0.2


result = (
    study_hours * study_weight
    + attendance * attendance_weight
    + homework_score * homework_weight
)

print (result)
```

# In summary, a neuron receives inputs, multiplies them by their respective weights, and then, add them together. After that, the activation function decides how strongly the neuron responds.
