# NEURAL NETWORK INTUITION

## Input - the information/clues given to network. This is exactly like the features learnt in machine learning

## Neuron like a worker 
## Layer - like a group of workers. This layer is into 3 sections
## Input Layer, Hidden Layer, Output Layer
## Input Layer - Where the clues enter
## Hidden Layer - Where clues are mixed together and useful patterns are learnt
## Output Layer - This is where the network gives its final answer. [ like prediction ]
## Weight - This mean the importance a network gives a clue. 
## Activation function - It helps the neuron to decide how strongly it should respond. Activation function = decides how strongly a neuron should respond after receiving and combining information.
## Gradient Descent - It's a method to gradually adjust the model so its error becomes smaller.
## Back propagation - It sends information about the mistake backwards so the network can know what needs adjusting.
