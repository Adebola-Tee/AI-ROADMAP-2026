```python
import pandas as pd
```

# Exercise 1 --------- Study Hours --> Score

```python
data = {
    "study_hours": [1, 2, 3, 4, 5, 6],
    "score": [40, 48, 55, 63, 70, 78]
}
df=pd.DataFrame(data)
df
```

```python
X = df[["study_hours"]]
y= df[["score"]]
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.linear_model import LinearRegression
model= LinearRegression()
model.fit(X,y)
```

```python
new_student = pd.DataFrame({
    "study_hours": [7]
})
prediction = model.predict(new_student)
prediction
```
