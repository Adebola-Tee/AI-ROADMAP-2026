```python
import pandas as pd
import numpy as np
```

## Exercise 1 - Find and count the missing values



```python
data = {
    "student": ["A", "B", "C", "D", "E"],
    "study_hours": [2, np.nan, 5, 6, np.nan],
    "score": [50, 60, np.nan, 80, 90]
}

df = pd.DataFrame(data)
df

```

```python
df.isnull()
```

```python
df.isnull().sum()
```

## Exercise 2 - Fill with mean

```python
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].mean())
df["study_hours"]
```

```python
df["score"]=df["score"].fillna(df["score"].mean())
df["score"]
```

```python
df
```

```python
df.isnull().sum()
```

## Exercise 3 - Learn with Median

```python
data = {
    "income": [20000, 22000, 23000, 25000, 200000, np.nan]
}
data_income = pd.DataFrame(data)
data_income
```

```python
data_income["income"]=data_income["income"].fillna(data_income["income"].median())
data_income["income"]
```

```python
data_income["income"]=data_income["income"].fillna(data_income["income"].mean())
data_income
```

## Exercise 4 - Missing text/category data

```python
data = {
    "student": ["A", "B", "C", "D", "E"],
    "subject": ["Maths", "English", np.nan, "Maths", "Maths"]
}

subject_df = pd.DataFrame(data)
subject_df
```

```python
subject_df["subject"]= subject_df["subject"].fillna(subject_df["subject"].mode()[0])
subject_df["subject"]
```

## Exercise 5 - Decide: Fill or delete?

```python
data = {
    "student": ["A", "B", "C", "D", "E"],
    "attendance": [90, np.nan, 75, np.nan, 88],
    "score": [80, 70, np.nan, 65, 85]
}

practice_df = pd.DataFrame(data)
practice_df
```

```python
practice_df.isnull().sum()
```

```python
remove_df=practice_df.dropna()
remove_df
```

```python
fill_df= practice_df.copy()
fill_df
```

```python
fill_df["attendance"]= fill_df["attendance"].fillna(fill_df["attendance"].mean())
fill_df
```

```python
fill_df["score"] = fill_df["score"].fillna(fill_df["score"].mean())
fill_df
```
