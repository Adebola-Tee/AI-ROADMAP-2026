```python
import pandas as pd
```

```python
data = {
    "student": ["Student A", "Student B", "Student C", "Student D", "Student E"],
    "weekly_practice_hours": [1, 2, 4, 5, 7],
    "attendance": [60, 75, 85, 95, 90],
    "actual_test_score": [45, 58, 74, 88, 86],
    "predicted_score": [50, 60, 72, 84, 90]
}
df = pd.DataFrame(data)
df
```

```python
df["error"] = df["actual_test_score"] - df["predicted_score"]
df
```

```python
df["absolute_error"] = df["error"].abs()
df
```
